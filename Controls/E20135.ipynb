# Derivation: Twin Rotor System Model

## Setup and Assumptions

Consider a rigid body (the twin rotor platform) constrained so that it **cannot rotate about its $\mathbf{b}_2$ body axis**. The body has:
- Inertia tensor $\mathbb{I}$ in the body frame
- Two rotors producing thrust forces at angles $\alpha$ and $\beta$ from $\mathbf{b}_1$, mounted in the $\mathbf{b}_1$–$\mathbf{b}_3$ plane
- Angular momentum $\pi = \mathbb{I}^R \omega$ in the spatial frame, where $\mathbb{I}^R = R\mathbb{I}R^T$

---

## Step 1: Kinematics on SO(3)

The orientation $R \in SO(3)$ satisfies the standard kinematic equation. For a rigid body with spatial angular velocity $\omega$, the time derivative of the rotation matrix is:

$$\boxed{\dot{R} = \widehat{\omega}R}$$

where $\widehat{\omega}$ is the $3\times 3$ skew-symmetric matrix such that $\widehat{\omega}v = \omega \times v$ for any vector $v$. This is the standard result from differentiating $RR^T = I$.

---

## Step 2: Angular Momentum Dynamics

Newton–Euler in the spatial frame gives:

$$\dot{\pi} = \tau_{\text{total}}$$

Splitting all applied torques into the constraint part $\tau^e$ (reaction moment enforcing the $\mathbf{b}_2$ constraint) and the control part $\tau^u$ (from the rotors):

$$\boxed{\dot{\pi} = \tau^e + \tau^u}$$

The spatial angular velocity is recovered from $\pi = \mathbb{I}^R\omega$, hence $\omega = (\mathbb{I}^R)^{-1}\pi$.

---

## Step 3: Deriving the Control Torque $\tau^u$

Each rotor $i$ produces a scalar thrust $u_i$ directed at a fixed angle in the $\mathbf{b}_1$–$\mathbf{b}_3$ plane of the body frame. In the body frame, the thrust directions are:

$$f_1^{\text{body}} = \begin{bmatrix}\cos\alpha \\ 0 \\ \sin\alpha\end{bmatrix}, \qquad f_2^{\text{body}} = \begin{bmatrix}-\cos\beta \\ 0 \\ -\sin\beta\end{bmatrix}$$

The signs reflect the counter-acting geometry of a twin-rotor (the two rotors apply moments in opposing senses). These forces produce torques acting on the body. Collecting both:

$$\tau^u_{\text{body}} = \begin{bmatrix}\cos\alpha & -\cos\beta \\ 0 & 0 \\ \sin\alpha & -\sin\beta\end{bmatrix}\begin{bmatrix}u_1 \\ u_2\end{bmatrix} = \underbrace{\begin{bmatrix}1&0\\0&0\\0&1\end{bmatrix}}_{3\times 2}\underbrace{\begin{bmatrix}\cos\alpha & -\cos\beta \\ \sin\alpha & -\sin\beta\end{bmatrix}}_{2\times 2}\begin{bmatrix}u_1\\u_2\end{bmatrix}$$

The selection matrix explicitly zeros out any direct actuation along $\mathbf{b}_2$, consistent with the rotor geometry. Rotating to the spatial frame:

$$\boxed{\tau^u = R\begin{bmatrix}1&0\\0&0\\0&1\end{bmatrix}\begin{bmatrix}\cos\alpha & -\cos\beta\\\sin\alpha&-\sin\beta\end{bmatrix}\begin{bmatrix}u_1\\u_2\end{bmatrix}}$$

---

## Step 4: Deriving the Constraint Torque $\tau^e$

The mechanical constraint prevents rotation about $\mathbf{b}_2$, i.e.:

$$e_2^T \Omega = 0 \quad \text{and} \quad e_2^T \dot{\Omega} = 0 \quad \text{for all time}$$

where $\Omega = R^T\omega$ is the body-frame angular velocity. The body-frame Euler equation is obtained by transforming $\dot{\pi} = \tau^e + \tau^u$ into the body frame. Using $\pi = R\,\mathbb{I}\,\Omega$ and differentiating:

$$\dot{\pi} = \dot{R}\,\mathbb{I}\,\Omega + R\,\mathbb{I}\,\dot{\Omega} = \widehat{\omega}R\,\mathbb{I}\,\Omega + R\,\mathbb{I}\,\dot{\Omega}$$

Pre-multiplying by $R^T$ and using $R^T\widehat{\omega}R = \widehat{\Omega}$ (the adjoint identity):

$$\mathbb{I}\dot{\Omega} + \widehat{\Omega}\,\mathbb{I}\,\Omega = R^T(\tau^e + \tau^u)$$

Recognising $\widehat{\Omega}\,\mathbb{I}\,\Omega = \Omega \times \mathbb{I}\Omega$, the full body-frame equation is:

$$\mathbb{I}\dot{\Omega} + \Omega\times\mathbb{I}\Omega = R^T\tau^e + R^T\tau^u$$

Since the rotors produce **no torque along $\mathbf{b}_2$** (the selection matrix zeros that row), we have $e_2^T R^T \tau^u = 0$. Projecting the equation onto $\mathbf{b}_2$ via $e_2^T$:

$$e_2^T\!\left(\mathbb{I}\dot{\Omega} + \Omega\times\mathbb{I}\Omega\right) = e_2^T R^T \tau^e$$

The constraint torque acts purely along $\mathbf{b}_2$ in the body frame, so $R^T\tau^e = \begin{bmatrix}0 & T_2 & 0\end{bmatrix}^T$, giving $e_2^T R^T\tau^e = T_2$. Therefore:

$$\boxed{T_2 = e_2^T\!\left(\Omega\times\mathbb{I}\Omega + \mathbb{I}\dot{\Omega}\right)}$$

and rotating back to the spatial frame:

$$\boxed{\tau^e = R\begin{bmatrix}0\\T_2\\0\end{bmatrix}}$$


# <font color="red">Task-2- Simulation of the Motion of the System for Different Inputs

Expressed in the body frame $\mathbf{b}$, the quantities $\mathbb{I}_1,\, \mathbb{I}_2,\,
\mathbb{I}_3$ represent the moments of inertia about the pitch, roll, and yaw axes,
respectively. Given that rolling motion is mechanically constrained in the experimental
setup, the parameter $\mathbb{I}_2$ plays no active role. Furthermore, $\mathbb{I}_3 >
\mathbb{I}_1$, which reflects the asymmetry in the physical configuration: yaw rotation
involves the additional mass of the upright support structure, whereas pitch rotation
engages only the connecting beam between the two propellers.

```python
import numpy as np
import plotly.graph_objects as go
from scipy.integrate import solve_ivp

# System Parameters
# Diagonal components of the inertia tensor (hypothetical values)
I1, I2, I3 = 1.0, 2.0, 1.5

# Input Configuration
def get_current_inputs(t):
    """
    Returns the current control inputs at time t.
    Activate the desired test case by uncommenting the corresponding line.
    """
    # input_array = [0.5 * np.sin(2 * t), -0.2 * np.cos(t)]  # Sinusoidal (Frequency Response)
    # input_array = [1.0, 1.0]                                 # Step Response
    # input_array = [1.0, 0.0]                                 # Pitch Axis Only
    input_array = [0.0, 1.0]                                   # Yaw Axis Only

    return input_array

# Actuator Torques
# Note: These are the torque inputs to be identified experimentally.

def u1_torque(t):
    """Returns the torque input for actuator 1."""
    return get_current_inputs(t)[0]

def u2_torque(t):
    """Returns the torque input for actuator 2."""
    return get_current_inputs(t)[1]
```

**Dynamics of the Twin Rotor System**

The following quantities are defined:

$$T_e = \begin{bmatrix} 0 \\ T_{e_2}\\ 0 \end{bmatrix}, \qquad
\Omega = \begin{bmatrix} \Omega_1 \\ \Omega_2 \\ \Omega_3 \end{bmatrix}, \qquad
\mathbb{I} = \begin{bmatrix} I_1 & 0 & 0 \\ 0 & I_2 & 0 \\ 0 & 0 & I_3 \end{bmatrix}$$

Applying the Euler rigid body equation:

$$
T_e + T_u = \mathbb{I}\dot{\Omega} + \Omega \times \mathbb{I}\Omega
$$

$$
\begin{bmatrix} 0 \\ T_{e_2}\\ 0 \end{bmatrix}
+ \begin{bmatrix} u_1\\ 0\\ -u_2 \end{bmatrix}
= \begin{bmatrix}
(I_3 - I_2)\,\Omega_2\Omega_3 + I_1\dot{\Omega}_1\\
(I_1 - I_3)\,\Omega_1\Omega_3 + I_2\dot{\Omega}_2\\
(I_2 - I_1)\,\Omega_1\Omega_2 + I_3\dot{\Omega}_3
\end{bmatrix}
$$

The angular velocity vector expressed in terms of Euler angles is:

$$
\Omega = \begin{bmatrix}
\dot{\phi} \\
\dot{\theta} \sin{\phi} \\
\dot{\theta} \cos{\phi}
\end{bmatrix}
$$

Taking the time derivative:

$$
\dot{\Omega} = \begin{bmatrix}
\ddot{\phi} \\
\ddot{\theta} \sin{\phi} + \dot{\theta}\,\dot{\phi} \cos{\phi} \\
\ddot{\theta} \cos{\phi} - \dot{\theta}\,\dot{\phi} \sin{\phi}
\end{bmatrix}
$$

Substituting into the Euler rigid body equation and expanding each component:

$$
u_1 = (I_3 - I_2)\,\dot{\theta}^2 \sin{\phi} \cos{\phi} + I_1\ddot{\phi}
$$

$$
T_{e_2} = (I_1 - I_3)\,\dot{\phi}\,\dot{\theta} \cos{\phi}
+ I_2\left(\ddot{\theta} \sin{\phi} + \dot{\theta}\,\dot{\phi} \cos{\phi}\right)
$$

$$
-u_2 = (I_2 - I_1)\,\dot{\phi}\,\dot{\theta} \sin{\phi}
+ I_3\left(\ddot{\theta} \cos{\phi} - \dot{\theta}\,\dot{\phi} \sin{\phi}\right)
$$

where $T_{e_2}$ denotes the reaction torque at the mechanical joint,
representing a constraint that must be accounted for in the design.

Rearranging, the governing equations of motion are obtained as:

$$
\ddot{\phi} = \frac{1}{I_1}\left[u_1 - (I_3 - I_2)\,\dot{\theta}^2 \sin{\phi} \cos{\phi}\right]
$$

$$
\ddot{\theta} = \frac{1}{\cos{\phi}}\left[\frac{1}{I_3}
\left[(I_1 - I_2)\,\dot{\phi}\,\dot{\theta} \sin{\phi} - u_2\right]
+ \dot{\theta}\,\dot{\phi} \sin{\phi}\right]
$$

It is worth noting that these equations become singular when $\phi = 90^\circ$,
a phenomenon known as **Gimbal Lock**. A controller built upon this Euler angle
representation would therefore be susceptible to numerical breakdown at this
configuration. To circumvent this limitation, the subsequent task reformulates
the control problem directly on $SO(3)$, bypassing the use of $\phi$ and $\theta$
entirely in favour of a coordinate-free error dynamics framework.

```python
# Differential Equations
def twin_rotor_dynamics(t, state):
    theta, phi, dtheta, dphi = state

    u1 = u1_torque(t)
    u2 = u2_torque(t)

    # Avoid divide-by-zero at gimbal lock (phi approaches 90 deg)
    cos_phi = np.cos(phi)
    if abs(cos_phi) < 1e-5:
        cos_phi = 1e-5 * np.sign(cos_phi) if cos_phi != 0 else 1e-5

    # Equations of motion derived from the Euler rigid body equations
    ddphi   = (u1 - (I3 - I2) * (dtheta**2) * np.sin(phi) * np.cos(phi)) / I1
    ddtheta = (-u2 - (I2 - I1 - I3) * dtheta * dphi * np.sin(phi)) / (I3 * cos_phi)

    return [dtheta, dphi, ddtheta, ddphi]


# Simulation Setup
t_span = (0, 10)
t_eval = np.linspace(t_span[0], t_span[1], 300)

# Initial conditions: [theta, phi, theta_dot, phi_dot]
# Small initial pitch offset with all velocities at rest
y0 = [0.1, 0.0, 0.0, 0.0]

# Integrate using 4th/5th order Runge-Kutta
solution = solve_ivp(twin_rotor_dynamics, t_span, y0, t_eval=t_eval, method='RK45')

theta_vals = solution.y[0]
phi_vals   = solution.y[1]


# Kinematics and 3D Geometry
def get_rotation_matrix(theta, phi):
    # Rotation about Z-axis (yaw)
    R3 = np.array([
        [np.cos(theta), -np.sin(theta), 0],
        [np.sin(theta),  np.cos(theta), 0],
        [0,              0,             1]
    ])
    # Rotation about X-axis (pitch)
    R1 = np.array([
        [1, 0,            0           ],
        [0, np.cos(phi), -np.sin(phi) ],
        [0, np.sin(phi),  np.cos(phi) ]
    ])
    return R3 @ R1

frames = []
L_support = 2.0  # Upright support length
L_beam    = 3.0  # Connecting beam length between the two rotors

for i in range(len(t_eval)):
    th = theta_vals[i]
    ph = phi_vals[i]
    R  = get_rotation_matrix(th, ph)

    # Upright support: fixed to intermediate frame, rotates with yaw only
    support_line = np.array([[0, 0, 0], [0, 0, -L_support]])

    # Connecting beam: aligned along the b2 body axis
    beam_local  = np.array([[0, -L_beam/2, 0], [0, L_beam/2, 0]])
    beam_global = (R @ beam_local.T).T

    # Rotor indicator lines at each beam end (along b1 and b3 directions)
    fan1_local  = np.array([[ 0.5, -L_beam/2, 0], [-0.5, -L_beam/2,  0  ]])
    fan2_local  = np.array([[ 0,    L_beam/2, 0.5], [0,   L_beam/2, -0.5]])

    fan1_global = (R @ fan1_local.T).T
    fan2_global = (R @ fan2_local.T).T

    # Combine segments with None separators for efficient Plotly line rendering
    x = [support_line[0,0], support_line[1,0], None,
         beam_global[0,0],  beam_global[1,0],  None,
         fan1_global[0,0],  fan1_global[1,0],  None,
         fan2_global[0,0],  fan2_global[1,0]]

    y = [support_line[0,1], support_line[1,1], None,
         beam_global[0,1],  beam_global[1,1],  None,
         fan1_global[0,1],  fan1_global[1,1],  None,
         fan2_global[0,1],  fan2_global[1,1]]

    z = [support_line[0,2], support_line[1,2], None,
         beam_global[0,2],  beam_global[1,2],  None,
         fan1_global[0,2],  fan1_global[1,2],  None,
         fan2_global[0,2],  fan2_global[1,2]]

    frames.append(go.Frame(data=[go.Scatter3d(x=x, y=y, z=z)]))


# Animation
fig = go.Figure(
    data=[go.Scatter3d(
        x=frames[0].data[0].x,
        y=frames[0].data[0].y,
        z=frames[0].data[0].z,
        mode='lines',
        line=dict(color='blue', width=6)
    )],
    layout=go.Layout(
        title="Twin Rotor System — Simulated Dynamics",
        scene=dict(
            xaxis=dict(range=[-3, 3], title='X (Earth Frame)'),
            yaxis=dict(range=[-3, 3], title='Y (Earth Frame)'),
            zaxis=dict(range=[-3, 3], title='Z (Earth Frame)'),
            aspectmode='cube'
        ),
        updatemenus=[dict(
            type="buttons",
            buttons=[dict(
                label="Play",
                method="animate",
                args=[None, dict(frame=dict(duration=20, redraw=True), fromcurrent=True)]
            )]
        )]
    ),
    frames=frames
)

fig.show()
```

# <font color="red">Task-3: Derivation and Simulation of a PID Controller for the Tracking Problem using Error Dynamics

Let $(o_r, R_r) \in SE(3)$ denote the reference configuration of the twin rotor system.
Since the centre of mass remains fixed at the pivot point, we have $o_r \equiv o$,
which allows us to restrict attention to the rotational dynamics governed by $R$.

To achieve asymptotic tracking of a sufficiently smooth desired trajectory $R_r$,
the configuration error $R_e \in SO(3)$ is introduced as follows.

**Body-frame (left-invariant) error:**

$$
R R_e = R_r
$$
$$
R_e = R^T R_r
$$

**Earth-frame (right-invariant) error:**

$$
R_e R = R_r
$$
$$
R_e = R_r R^T
$$

As the equations of motion are expressed in the $\mathbf{e}$ frame,
the right-invariant formulation is adopted for all subsequent derivations.

**Part I: Derivation of Tracking Error Dynamics**

Let $\omega_r$ denote the spatial angular velocity associated with the desired trajectory,
defined through the relation $\widehat{\omega}_r = \dot{R}_r R_r^T$.

The **spatial angular velocity error** is defined as:

$$
\widehat{\omega}_e \triangleq \dot{R}_e R_e^T
$$

Substituting $R_e = R_r R^T$ and differentiating:

$$
\widehat{\omega}_e = \left(\dot{R}_r R^T + R_r \dot{R}^T\right) R_e^T
$$

$$
\widehat{\omega}_e = \dot{R}_r\left(R^T R_e^T\right) + \left(R_e R\right)\dot{R}^T R_e^T
$$

$$
\widehat{\omega}_e = \dot{R}_r R_r^T + R_e \left(R \dot{R}^T\right) R_e^T
$$

Noting that $RR^T = I$ implies $R\dot{R}^T + \dot{R}R^T = 0$,
and therefore $R\dot{R}^T = -\dot{R}R^T$:

$$
\widehat{\omega}_e = \dot{R}_r R_r^T + R_e \left(-\dot{R} R^T\right) R_e^T
$$

$$
\widehat{\omega}_e = \widehat{\omega}_r - R_e \widehat{\omega}\, R_e^T
$$

Applying the rotation identity $R_e \widehat{\omega}\, R_e^T = \widehat{R_e \omega}$:

$$
\widehat{\omega}_e = \widehat{\omega}_r - \widehat{R_e \omega}
$$

Extracting the corresponding vectors from both sides yields:

$$
\omega_e = \omega_r - R_e \omega
$$

For the **error in spatial angular momentum**,

In line with the system define the reference spatial angular momentum $\pi_r\triangleq R_r\mathbb{I}R_r^T\omega_r$. Then, by defining the error in spatial angular momentum to be:

$$
\pi_e\triangleq R\mathbb{I}R_r^T\omega_e$$

$$
\pi_e = R\mathbb{I}R_r^T(\omega_r-R_e\omega) $$

$$
\pi_e = (R_e^T R_r)\mathbb{I}R_r^T\omega_r-R\mathbb{I}R^T\omega $$

$$
\pi_e =R_e^T\pi_r-\pi $$

Differentiating w.r.t time:

$$
\dot{\pi}_e = \dot{R_e^T}\pi_r + R_e^T\dot{\pi}_r - \dot{\pi} $$

$$
\dot{\pi}_e = ({R}_e^T \widehat{\omega}_e^T)\pi_r + R_e^T\dot{\pi}_r - \dot{\pi} $$

$$
\dot{\pi}_e = {R}_e^T (-\widehat{\omega}_e)\pi_r + R_e^T\dot{\pi}_r - \dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(\dot{\pi}_r -\widehat{\omega}_e\pi_r)-\dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(\dot{\pi}_r-\omega_e\times \pi_r)-\dot{\pi} $$

From $\pi_r = R_r \Pi_r$,

$$
\dot{\pi}_e = R_e^T(\dot{R}_r \Pi_r + R_r \dot{\Pi}_r-\omega_e\times \pi_r)-\dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(\widehat{\omega}_r{R}_r \Pi_r + R_r \dot{\Pi}_r-\omega_e\times \pi_r)-\dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(R_r \dot{\Pi}_r+ \widehat{\omega}_r \pi_r-\omega_e\times \pi_r)-\dot{\pi} $$

$$
\dot{\pi}_e = R_e^T(R_r\dot{\Pi}_r+(\omega_r-\omega_e)\times \pi_r)-\dot{\pi}$$

$$
\dot{\pi}_e = R_e^T(R_r\dot{\Pi}_r+ R_e\omega\times \pi_r)-\dot{\pi}
$$

$$
\dot{\pi}_e = (R\dot{\Pi}_r+ \omega\times R_e^T\pi_r)-\dot{\pi}
$$

Thus, we have the error dynamics:
$$
\dot{R}_e=\widehat{\omega}_eR_e
$$
$$
\dot{\pi}_e=(R\dot{\Pi}_r+\omega\times R_e^T\pi_r)-\tau_u-\tau_e
$$

**Part II: Lyapunov Stability Analysis**

As established earlier:

$$
R_e = R_r R^T \in SO(3)
$$

Observe that as $R \rightarrow R_r$, the error naturally tends to the identity,
i.e., $R_e \rightarrow I$. To quantify this deviation, a potential energy function
$\mathcal{V}(R_e) \in \mathbb{R}_{\ge 0}$ is constructed such that it attains its
minimum value $\mathcal{V}(R_e) \rightarrow 0$ at the equilibrium $R_e = I$:

$$
\mathcal{V}(R_e) = \frac{1}{2} \operatorname{trace}\!\left(K(I - R_e)\right)
$$

Here, $K$ is an inertia-weighted scaling matrix defined as:

$$
K = \begin{bmatrix}
\mathbb{I}_2 + \mathbb{I}_3 - \mathbb{I}_1 & 0 & 0 \\
0 & \mathbb{I}_1 + \mathbb{I}_3 - \mathbb{I}_2 & 0 \\
0 & 0 & \mathbb{I}_1 + \mathbb{I}_2 - \mathbb{I}_3
\end{bmatrix}
$$

Then, the Lyapunov function $\mathcal{W}(R_e, \Omega_e) \in \mathbb{R}_{\ge 0}$ can be written as:

$$
\mathcal{W}(R_e, \Omega_e) = \underbrace{\frac{1}{2} \text{trace}(K(I-R_e))}_{\text{Potential Energy}} + \underbrace{\frac{1}{2} \langle \mathbb{I}\Omega_e, \Omega_e \rangle}_{\text{Kinetic Energy}}
$$

To find **rate of change of potential energy**,

$$
\mathcal{V}(R_e) = \frac{1}{2} \text{trace}(K(I-R_e))
$$
Differentiating w.r.t. time:
$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2}\text{trace}(K \dot{R}_e)
$$
From $\dot{R}_e = \widehat{\omega}_eR_e$,
$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2}\text{trace}(K \widehat{\omega}_eR_e)
$$

By cyclic permutation of trace of a product,

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2}\text{trace}(\widehat{\omega}_eR_eK)
$$
$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2}\text{trace}((\widehat{\omega}_eR_eK)^T)
$$
$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{2}\text{trace}(K^T R_e^T \widehat{\omega}_e^T)
$$

By symmetry, $K = K^T$ and by skew-symmetry $\widehat{\omega}_e^T = -\widehat{\omega}_e$

$$
\dot{\mathcal{V}}(R_e) = \frac{1}{2}\text{trace}(K R_e^T \widehat{\omega}_e)
$$
$$
\dot{\mathcal{V}}(R_e) = \frac{1}{2}\text{trace}(\widehat{\omega}_e K R_e^T)
$$

From $\text{trace}(\widehat{\omega}_e K R_e^T) = -\text{trace}(\widehat{\omega}_eR_eK)$,

$$
\dot{\mathcal{V}}(R_e) = \frac{1}{2}\left(\frac{(-\text{trace}(\widehat{\omega}_eR_eK))+\text{trace}(\widehat{\omega}_e K R_e^T)}{2}\right)
$$

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{4}\text{trace}(\widehat{\omega}_e(R_eK-KR_e^T))
$$

From the property $\text{trace}(\hat{x}\hat{y}) = -2x^Ty$ and since $(R_eK-KR_e^T) \in SO(3)$,

$$
\dot{\mathcal{V}}(R_e) = -\frac{1}{4}(-2\omega_e^T)(R_eK-KR_e^T)^{\vee}
$$

where $\vee: SO(3) \rightarrow \mathbb{R}^3$ (inverse of hat operation)

$$
\dot{\mathcal{V}}(R_e) = \omega_e^T\underbrace{\frac{1}{2}(R_eK-KR_e^T)^{\vee}}_{\nabla \mathcal{V}}
$$

$$
\dot{\mathcal{V}}(R_e) = \omega_e^T\nabla \mathcal{V}
$$


$\nabla \mathcal{V} \in \mathbb{R}^3$ is the rate of change of potential energy with rotation, i.e. spatial derivative.

At equilibrium, $\dot{\mathcal{V}} = 0$ requires $\nabla \mathcal{V} = 0$ for all
$\omega_e \in \mathbb{R}^3$, which leads to the condition:

$$
(R_e K - K R_e^T) = 0 \quad \forall \; \omega_e
$$

Since $K$ is a positive definite diagonal matrix, attention is restricted to the
subset $R'_e \in \{\, R_e \mid \det(R_e) = 1,\; R_e \text{ orthogonal and diagonal}\,\}$,
for which:

$$
\dot{\mathcal{V}}(R'_e) = 0 \quad \forall \; \omega_e
$$

The complete set of rotation matrices satisfying these conditions is:

$$
R'_e = \left\{
\begin{bmatrix} 1&0&0\\ 0&1&0\\ 0&0&1 \end{bmatrix},\;
\begin{bmatrix} 1&0&0\\ 0&-1&0\\ 0&0&-1 \end{bmatrix},\;
\begin{bmatrix} -1&0&0\\ 0&1&0\\ 0&0&-1 \end{bmatrix},\;
\begin{bmatrix} -1&0&0\\ 0&-1&0\\ 0&0&1 \end{bmatrix}
\right\}
$$

Evaluating both $\mathcal{V}$ and $\dot{\mathcal{V}}$ across each candidate yields:

- $R_e = \operatorname{diag}(1,1,1)$: $\;\mathcal{V} = 0$ — a **global minimum** (stable equilibrium)
- $R_e = \operatorname{diag}(1,-1,-1)$: $\;\mathcal{V} = 2\mathbb{I}_1 \neq 0$ — a **local maximum** (unstable)
- $R_e = \operatorname{diag}(-1,1,-1)$: $\;\mathcal{V} = 2\mathbb{I}_2 \neq 0$ — a **local maximum** (unstable)
- $R_e = \operatorname{diag}(-1,-1,1)$: $\;\mathcal{V} = 2\mathbb{I}_3 \neq 0$ — a **local maximum** (unstable)

These results confirm that $R_e = I$ is the unique globally stable equilibrium,
while the remaining three critical points are unstable maxima.

Each unstable critical point corresponds to an **antipodal configuration** — a
rotation of exactly $180^\circ$ about one of the three principal axes relative to
the reference orientation. Any infinitesimal perturbation is sufficient to drive
the system away from these points toward the stable equilibrium. Consequently,
in the absence of a reference velocity $(\omega_r = 0)$, the actuators produce no
corrective torque at these configurations despite a non-zero position error
$(\mathcal{V} \neq 0)$. In practice, however, system noise ensures that the twin
rotor never remains trapped at such unstable points.

To find **rate of change of kinetic energy**,

$$
\mathcal{K}(\Omega_e) = \frac{1}{2} \langle \mathbb{I}\Omega_e, \Omega_e \rangle
$$

By definition,

$$
\omega_e = R_e \Omega_e \,\,\,\,\text{and}\,\,\,\, \pi_e = R_e\mathbb{I}\Omega_e
$$

Thus,

$$
\langle \pi_e, \omega_e \rangle = \pi_e^T \omega_e = (R_e\mathbb{I}\Omega_e)^T(R_e\Omega_e) = \Omega_e^T\mathbb{I}^TR_e^TR_e\Omega_e = \Omega_e^T\mathbb{I}^T\Omega_e =  \langle \mathbb{I}\Omega_e, \Omega_e \rangle
$$

Then,

$$
\mathcal{K}(\Omega_e) = \frac{1}{2} \langle \mathbb{I}\Omega_e, \Omega_e \rangle = \frac{1}{2} \langle \pi_e, \omega_e \rangle
$$

This proves that the kinetic energy in body frame is identical to that of Earth frame. Rotation matrix $R_e$ preserves the kinetic energy regardless of the inertial frame.

$$
\mathcal{K}(\Omega_e) = \frac{1}{2} \langle \mathbb{I}\Omega_e, \Omega_e \rangle
$$

Differentiating w.r.t. time:

$$
\dot{\mathcal{K}}(\Omega_e) = \frac{1}{2} [\langle \mathbb{I}\dot{\Omega}_e, \Omega_e \rangle + \langle \mathbb{I}{\Omega}_e, \dot{\Omega}_e \rangle]
$$

From $\langle \mathbb{I}{\Omega}_e, \dot{\Omega}_e \rangle = (\mathbb{I}{\Omega}_e)^T\dot{\Omega}_e = \Omega_e^T \mathbb{I}^T \dot{\Omega}_e = \Omega_e^T(\mathbb{I} \dot{\Omega}_e) = \langle \mathbb{I}\dot{\Omega}_e, \Omega_e \rangle $,

$$
\dot{\mathcal{K}}(\Omega_e) = \frac{1}{2} [\langle \mathbb{I}\dot{\Omega}_e, \Omega_e \rangle + \langle \mathbb{I}\dot{\Omega}_e, \Omega_e \rangle]
$$

$$
\dot{\mathcal{K}}(\Omega_e) = \langle \mathbb{I}\dot{\Omega}_e, \Omega_e \rangle
$$

$$
\dot{\mathcal{K}}(\Omega_e) = \Omega_e^T \mathbb{I}\dot{\Omega}_e
$$

Then, the Lyapunov function can be written as,

$$\mathcal{W}(R_e,\Omega_e) = \mathcal{V}(R_e) + \mathcal{K}(\Omega_e) \ge 0
$$

Differentiating w.r.t. time:

$$\dot{\mathcal{W}}(R_e,\Omega_e) = \dot{\mathcal{V}}(R_e) + \dot{\mathcal{K}}(\Omega_e)
$$

$$\dot{\mathcal{W}}(R_e,\Omega_e) = \omega_e^T\nabla \mathcal{V} + \Omega_e^T \mathbb{I}\dot{\Omega}_e
$$

$$\dot{\mathcal{W}}(R_e,\Omega_e) = (R_e\Omega_e)^T\nabla \mathcal{V} + \Omega_e^T \mathbb{I}\dot{\Omega}_e
$$

$$\dot{\mathcal{W}}(R_e,\Omega_e) = \Omega_e^T R_e^T \nabla \mathcal{V} + \Omega_e^T \mathbb{I}\dot{\Omega}_e
$$

$$\dot{\mathcal{W}}(R_e,\Omega_e) = \Omega_e^T (R_e^T \nabla \mathcal{V} + \mathbb{I}\dot{\Omega}_e)
$$

Recall:

$$ \omega_e = \omega_r - R_e \omega $$

$$ R_e\Omega_e = R_r\Omega_r - R_eR\Omega $$

$$ R_e\Omega_e = R_r(\Omega_r - \Omega) $$

$$ \Omega_e = R_e^TR_r(\Omega_r - \Omega) $$

$$ \Omega_e = R(\Omega_r - \Omega) $$

**Part III: Design of the PID Controller**

Starting from the error dynamic equation:

$$
\dot{\pi}_e = \left(R\dot{\Pi}_r + \omega \times R_e^T \pi_r\right) - \tau_u - \tau_e
$$

The actuator input is designed as:

$$
\tau_u = \underbrace{\left(R\dot{\Pi}_r + \omega \times R_e^T \pi_r\right)}_{\text{Feedforward}}
- \underbrace{k_P e_R - k_D e_{\omega} - k_I e_{I_R}}_{\text{Feedback}}
$$

where $e_R,\, e_{\omega},\, e_{I_R} \in \mathbb{R}^3$ are error terms to be determined,
and $k_P,\, k_D,\, k_I \in \mathbb{M}_{3 \times 3}$ are diagonal gain matrices of the form:

$$
k_i = \begin{bmatrix}
k_{i_1} & 0 & 0 \\
0 & 0 & 0 \\
0 & 0 & k_{i_3}
\end{bmatrix} \qquad \text{for} \quad i = P,\, I,\, D
$$

The independent entries $k_{i_1}$ and $k_{i_3}$ permit separate tuning of the pitch
and yaw channel weights, respectively. The middle diagonal entry vanishes
($k_{i_2} \rightarrow 0$) since the roll degree of freedom is mechanically constrained.

Substituting this control law back into the error dynamics, the nonlinear terms —
including gyroscopic effects and inertia coupling — cancel exactly, yielding:

$$
\dot{\pi}_e = k_P e_R + k_D e_{\omega} + k_I e_{I_R} - \tau_e
$$

The holonomic constraint torque $\tau_e$ is passively compensated by the mechanical
joints and may therefore be neglected, reducing the expression to:

$$
\dot{\pi}_e = k_P e_R + k_D e_{\omega} + k_I e_{I_R}
$$

The resulting error dynamics are predominantly linear, bearing close resemblance
to a conventional Euclidean PID controller structure.

**Proportional error term $e_R$:**

The proportional error is defined as the gradient of the potential function
$\mathcal{V}$, ensuring that as this gradient grows, the error signal drives
the system toward the minimum $\mathcal{V} = 0$ — that is, toward the desired
trajectory ($\dot{\mathcal{V}} < 0$):

$$
e_R = \nabla \mathcal{V} = \frac{1}{2}\left(KR_e^T - R_e K\right)^{\vee}
$$

**Derivative error term $e_{\omega}$:**

The derivative error is proportional to the angular momentum error, and acts to
suppress excessive angular velocity ($\dot{\mathcal{K}} < 0$), ensuring the system
converges to the desired trajectory without oscillations or overshoot:

$$
e_{\omega} = \pi_e = R_e^T \pi_r - \pi
$$

**Integral error term $e_{I_R}$:**

The integral term accumulates the orientation error over time, providing
corrections for persistent steady-state disturbances such as friction or
sustained aerodynamic forces:

$$
e_{I_R} = \int e_R \, dt
$$

The complete control law is given by:

$$
\tau_u = \left(R\dot{\Pi}_r + \omega \times R_e^T\pi_r\right) - k_P e_R - k_D e_{\omega} - k_I e_{I_R}
$$

The feedforward component prescribes the torque required to counteract inertial
and gyroscopic effects along the desired trajectory, while the feedback component
corrects residual errors in orientation, angular velocity, and steady-state disturbances.

Since $\tau_u = RT_u = R \begin{bmatrix} u_1 & 0 & -u_2 \end{bmatrix}^T$,
pre-multiplying both sides by $R^T$ yields:

$$
\begin{bmatrix} u_1 \\ 0 \\ -u_2 \end{bmatrix} = R^T\!\left[
\left(R\dot{\Pi}_r + \omega \times R_e^T\pi_r\right)
- k_P e_R - k_D\left(R_e^T\pi_r - \pi\right) - k_I e_{I_R}
\right]
$$

Under the simplifying assumption $k_{i_1} = k_{i_3}$, the gain matrices reduce to
scalars $k_P,\, k_I,\, k_D \in \mathbb{R}$, and the expression simplifies to:

$$
\begin{bmatrix} u_1 \\ 0 \\ -u_2 \end{bmatrix} =
\left(\dot{\Pi}_r + \Omega \times \Pi_r\right)
- k_P R^T e_R - k_D\left(\Pi_r - \Pi\right) - k_I R^T e_{I_R}
$$

**Part IV: PID Controller Simulation**

```python
from scipy.spatial.transform import Rotation
from scipy import linalg

# Helper Utilities
class QQHelper:

    def r_from_quaternions(self, q):
        """
        Converts a unit quaternion to a rotation matrix.
        Input format: q = [qw, qx, qy, qz] (scalar-first convention).
        Scipy expects scalar-last format [qx, qy, qz, qw].
        """
        norm = np.linalg.norm(q)
        q = q / norm if norm > 1e-6 else [1, 0, 0, 0]
        return Rotation.from_quat([q[1], q[2], q[3], q[0]]).as_matrix()

    def hat_matrix(self, v):
        """
        Constructs the skew-symmetric (hat) matrix of a 3-vector v,
        such that hat(v) @ w = v x w for any vector w.
        """
        return np.array([
            [    0, -v[2],  v[1]],
            [ v[2],     0, -v[0]],
            [-v[1],  v[0],     0]
        ])

qq = QQHelper()


# PID Gain Tuning
kp = 8.0   # Proportional gain
kd = 4.0   # Derivative gain
ki = 0.0   # Integral gain (set to zero if steady-state correction is not required)


# Reference Trajectory Definition
def my_trajectory(t):
    """
    Returns the reference trajectory states at time t.
    Activate the desired example by uncommenting the corresponding block.
    """

    # Example 1: Sinusoidal pitch oscillation with constant yaw spin
    phi     = 0.2 * np.sin(t)
    dphi    = 0.2 * np.cos(t)
    ddphi   = -0.2 * np.sin(t)

    theta   = 0.5 * t
    dtheta  = 0.5
    ddtheta = 0.0

    # Example 2: Smooth step to a fixed orientation (0.5 rad pitch, static yaw)
    # phi     = 0.3 * np.sin(0.5 * t)
    # dphi    = 0.15 * np.cos(0.5 * t)
    # ddphi   = -0.075 * np.sin(0.5 * t)

    # theta   = 0.2 * np.pi
    # dtheta  = 0.0
    # ddtheta = 0.0

    return phi, dphi, ddphi, theta, dtheta, ddtheta


def referenceConfig(II, t, trajectory_func):
    """
    Computes the reference rotation matrix, angular momentum,
    its time derivative, and angular velocity at time t.
    """
    phi, dphi, ddphi, theta, dtheta, ddtheta = trajectory_func(t)

    # Guard against gimbal lock by clamping cos(phi) away from zero
    cos_phi = np.cos(phi)
    if abs(cos_phi) < 1e-5:
        sign    = np.sign(cos_phi) if cos_phi != 0 else 1.0
        cos_phi = 1e-5 * sign
        phi     = np.arccos(cos_phi)

    sin_phi   = np.sin(phi)
    cos_theta = np.cos(theta)
    sin_theta = np.sin(theta)

    # Reference rotation matrix: Rr = R3 (yaw) @ R1 (pitch)
    R3 = np.array([
        [cos_theta, -sin_theta, 0],
        [sin_theta,  cos_theta, 0],
        [        0,          0, 1]
    ])
    R1 = np.array([
        [1,       0,        0],
        [0, cos_phi, -sin_phi],
        [0, sin_phi,  cos_phi]
    ])
    Rr = R3 @ R1

    # Reference angular velocity in body frame
    Omegart = np.array([
        dphi,
        dtheta * sin_phi,
        dtheta * cos_phi
    ])

    # Reference angular acceleration in body frame
    dotOmegart = np.array([
        ddphi,
        ddtheta * sin_phi + dtheta * dphi * cos_phi,
        ddtheta * cos_phi - dtheta * dphi * sin_phi
    ])

    # Reference angular momentum and its time derivative
    Pir    = II @ Omegart
    dotPir = II @ dotOmegart + np.cross(Omegart, II @ Omegart)

    return [Rr, Pir, dotPir, Omegart]


# Controller
def controller(qq, parameters, t, X, taue, fe):
    """
    Computes the control torque using a geometric PID law on SO(3).
    Returns the Earth-frame torque, force, orientation error, and velocity error.
    """
    II    = parameters['II']
    R     = X[0][0]
    omega = X[1]
    piI   = X[3]

    Rr, Pir, dotPir, Omegart = referenceConfig(II, t, my_trajectory)

    # Configuration error in the Earth frame
    Re = Rr @ R.T

    # Inertia-weighted gain matrix K
    I1, I2, I3 = np.diag(II)
    K = np.diag([I2 + I3 - I1, I1 + I3 - I2, I1 + I2 - I3])

    # Proportional orientation error eR (vee of skew-symmetric part)
    eRhat = 0.5 * (Re @ K - K @ Re.T)
    eR    = np.array([-eRhat[1, 2], eRhat[0, 2], -eRhat[0, 1]])

    # Body-frame quantities
    Omega   = R.T @ omega      # Body angular velocity
    Pi      = II @ Omega       # Body angular momentum
    Pi_r    = II @ Omegart     # Reference body angular momentum
    dotPi_r = dotPir           # Reference momentum rate

    # Feedforward: compensates inertial and gyroscopic effects
    Tu_ff = dotPi_r + np.cross(Omega, Pi_r)

    # Feedback: PID correction in body frame
    Tu_fb = kp * (R.T @ eR) + kd * (Pi_r - Pi) + ki * (R.T @ piI)

    Tu_nom = Tu_ff + Tu_fb

    # Extract actuator inputs (roll channel is constrained to zero)
    u1     = Tu_nom[0]
    u2     = -Tu_nom[2]
    T_body = np.array([u1, 0, -u2])

    # Map body-frame torque back to Earth frame
    tauu = R @ T_body

    fu = -fe + np.array([0, 0, 0])
    return [tauu, fu, eR, (Pi_r - Pi)]


# Rigid Body Dynamics
def rigid_body_dynamics(t, state):
    """
    Defines the ODE system for the twin rotor rigid body.
    State vector: [qw, qx, qy, qz, Wx, Wy, Wz, piI_x, piI_y, piI_z]
    Note: omega represents the body-frame angular velocity.
    """
    q          = state[0:4]
    omega_body = state[4:7]
    piI        = state[7:10]

    # Normalise quaternion to prevent numerical drift
    q = q / np.linalg.norm(q)
    R = qq.r_from_quaternions(q)

    # Convert body-frame omega to spatial (Earth-frame) omega for the controller
    omega_spatial = R @ omega_body

    X = [[R], omega_spatial, None, piI]
    taue, fe = np.zeros(3), np.zeros(3)

    Uc   = controller(qq, parameters, t, X, taue, fe)
    tauu = Uc[0]   # Earth-frame control torque
    eR   = Uc[2]   # Orientation error for integral accumulation

    T_body = R.T @ tauu

    # Quaternion kinematics (uses body-frame angular velocity)
    q_dot = 0.5 * np.array([
        -q[1]*omega_body[0] - q[2]*omega_body[1] - q[3]*omega_body[2],
         q[0]*omega_body[0] - q[3]*omega_body[1] + q[2]*omega_body[2],
         q[3]*omega_body[0] + q[0]*omega_body[1] - q[1]*omega_body[2],
        -q[2]*omega_body[0] + q[1]*omega_body[1] + q[0]*omega_body[2]
    ])

    # Euler equations in the body frame
    II        = parameters['II']
    omega_dot = linalg.inv(II) @ (T_body - np.cross(omega_body, II @ omega_body))

    # Enforce holonomic constraint: roll is mechanically locked
    omega_dot[1] = 0.0
    piI_dot      = eR

    return np.concatenate((q_dot, omega_dot, piI_dot))


# Simulation Setup
t_span = (0, 30)
t_eval = np.linspace(t_span[0], t_span[1], 750)

parameters = {
    'II': np.diag([1.0, 2.0, 1.5])
}

# Initial state: [qw, qx, qy, qz, Wx, Wy, Wz, piI_x, piI_y, piI_z]
y0 = [0.8, 0.5, 0, 0.1, 0, 0, 0, 0, 0, 0]

solution = solve_ivp(rigid_body_dynamics, t_span, y0, t_eval=t_eval, method='RK45')


# Geometry and Animation
L_support, L_beam = 2.0, 3.0
frames = []

def build_geometry(R):
    """Constructs the beam and rotor indicator segments in the Earth frame."""
    beam = (R @ np.array([[0, -L_beam/2, 0], [0,  L_beam/2,    0]]).T).T
    fan1 = (R @ np.array([[0.5, -L_beam/2, 0], [-0.5, -L_beam/2,  0]]).T).T
    fan2 = (R @ np.array([[0,  L_beam/2,  0.5], [0,   L_beam/2, -0.5]]).T).T
    return beam, fan1, fan2

for i, t in enumerate(solution.t):
    q_act = solution.y[0:4, i]
    R_act = qq.r_from_quaternions(q_act)

    R_ref, Pir_ref, dotPir_ref, Omegart_ref = referenceConfig(
        parameters['II'], t, my_trajectory
    )

    beam_act, fan1_act, fan2_act = build_geometry(R_act)
    beam_ref, fan1_ref, fan2_ref = build_geometry(R_ref)
    support = np.array([[0, 0, 0], [0, 0, -L_support]])

    def extract_coords(support, beam, fan1, fan2):
        x = [support[0,0], support[1,0], None, beam[0,0], beam[1,0], None,
             fan1[0,0], fan1[1,0], None, fan2[0,0], fan2[1,0]]
        y = [support[0,1], support[1,1], None, beam[0,1], beam[1,1], None,
             fan1[0,1], fan1[1,1], None, fan2[0,1], fan2[1,1]]
        z = [support[0,2], support[1,2], None, beam[0,2], beam[1,2], None,
             fan1[0,2], fan1[1,2], None, fan2[0,2], fan2[1,2]]
        return x, y, z

    x_ref, y_ref, z_ref = extract_coords(support, beam_ref, fan1_ref, fan2_ref)
    x_act, y_act, z_act = extract_coords(support, beam_act, fan1_act, fan2_act)

    frames.append(go.Frame(data=[
        go.Scatter3d(
            x=x_ref, y=y_ref, z=z_ref,
            mode='lines', name="Reference",
            line=dict(color='rgba(255, 99, 71, 0.5)', width=8)
        ),
        go.Scatter3d(
            x=x_act, y=y_act, z=z_act,
            mode='lines', name="Twin Rotor",
            line=dict(color='blue', width=6)
        )
    ]))

fig = go.Figure(
    data=[frames[0].data[0], frames[0].data[1]],
    layout=go.Layout(
        title="Geometric PID Controller — Trajectory Tracking on SO(3)",
        scene=dict(
            xaxis=dict(range=[-3, 3], title='X (Earth Frame)'),
            yaxis=dict(range=[-3, 3], title='Y (Earth Frame)'),
            zaxis=dict(range=[-3, 3], title='Z (Earth Frame)'),
            aspectmode='cube'
        ),
        updatemenus=[dict(
            type="buttons",
            buttons=[dict(
                label="Play",
                method="animate",
                args=[None, dict(frame=dict(duration=20, redraw=True), fromcurrent=True)]
            )]
        )]
    ),
    frames=frames
)

fig.show()
```
